```python
import gradio as gr
import sympy as sp

from nlo import (
    calculate_gradient,
    build_lagrangian,
    find_lagrange_candidates,
    Lagrange_Equality
)


def optimize(fn, constraints, n_vars, optimization):

    try:
        if not fn.strip():
            raise ValueError("Enter an objective function.")

        constraint_list = [
            line.strip()
            for line in constraints.splitlines()
            if line.strip()
        ]

        if not constraint_list:
            raise ValueError("Enter at least one constraint.")

        if "=" in fn or any("=" in c for c in constraint_list):
            raise ValueError(
                "Enter expressions without the '=' sign."
            )

        n_vars = int(n_vars)

        if n_vars < 1:
            raise ValueError("Number of variables must be positive.")

        minimize = optimization == "Minimize"

        # Calculate gradient
        gradient = calculate_gradient(fn, n_vars)

        # Build Lagrangian
        lagrangian = build_lagrangian(
            fn, constraint_list, n_vars
        )

        # Find candidate points
        candidates = find_lagrange_candidates(
            fn, constraint_list, n_vars
        )

        # Find best solution
        best_point, best_value = Lagrange_Equality(
            fn,
            constraint_list,
            n_vars=n_vars,
            minimize=minimize
        )

        gradient_text = "\n".join(
            f"∂f/∂x_{i} = {derivative}"
            for i, derivative in enumerate(gradient)
        )

        candidate_text = "\n".join(
            str(point) for point in candidates
        )

        if best_point is None:
            result_text = "No isolated real feasible candidate found."
        else:
            result_text = (
                f"Best Point: {best_point}\n"
                f"Objective Value: {best_value}"
            )

        return (
            str(lagrangian),
            gradient_text,
            candidate_text or "No candidates found.",
            result_text
        )

    except Exception as error:
        raise gr.Error(str(error))


with gr.Blocks(title="NLO Lagrange Optimizer") as demo:

    gr.Markdown(
        """
        # NLO — Lagrange Optimizer

        ### Equality-Constrained Optimization
        Solve optimization problems using Lagrange Multipliers.
        """
    )

    with gr.Row():

        with gr.Column():

            objective = gr.Textbox(
                label="Objective Function",
                placeholder="x_0**2 + x_1**2",
                value="x_0**2 + x_1**2"
            )

            constraints = gr.Textbox(
                label="Equality Constraints",
                placeholder="One constraint per line",
                value="x_0 + x_1 - 2",
                lines=4
            )

            n_vars = gr.Number(
                label="Number of Variables",
                value=2,
                precision=0
            )

            optimization = gr.Radio(
                choices=["Minimize", "Maximize"],
                value="Minimize",
                label="Optimization Type"
            )

            solve_button = gr.Button(
                "Solve Optimization",
                variant="primary"
            )

        with gr.Column():

            lagrangian_output = gr.Textbox(
                label="Lagrangian Function"
            )

            gradient_output = gr.Textbox(
                label="Gradient"
            )

            candidates_output = gr.Textbox(
                label="Lagrange Candidates",
                lines=5
            )

            result_output = gr.Textbox(
                label="Optimization Result",
                lines=3
            )

    solve_button.click(
        fn=optimize,
        inputs=[
            objective,
            constraints,
            n_vars,
            optimization
        ],
        outputs=[
            lagrangian_output,
            gradient_output,
            candidates_output,
            result_output
        ]
    )

    gr.Markdown(
        """
        ---
        **Nonlinear Optimization — Fall 2026**
        Alexandria University | Faculty of Computers and Data Science
        """
    )


if __name__ == "__main__":
    demo.launch()
```
